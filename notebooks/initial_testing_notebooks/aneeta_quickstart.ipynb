```python
# 6) Load ANEETA Dataset → normalize to DataFrame (repo data only)
import pandas as pd, json, os, re
from pathlib import Path

# Local convenience path (optional)
local_processed = (ROOT / "datasets" / "aneeta" / "processed").resolve()
# Authoritative repo path
repo_processed = (aneeta_dir / "Processed Data").resolve()

rows = []
source_used = None

# 1) Prefer local test.jsonl if present (explicit test set)
test_path = local_processed / "test.jsonl"
if test_path.exists():
    import jsonlines
    with jsonlines.open(test_path, 'r') as reader:
        rows = list(reader)
    source_used = str(test_path)

# Helper: specialized parser for NEET solved papers (extract Q and numeric answer)
def parse_solved_mcqs(json_path: Path):
    try:
        data = json.loads(json_path.read_text(encoding='utf-8'))
    except Exception:
        return []
    # Concatenate pages and search for patterns like: "1. <question/body> ... Answer (3)"
    text = "\n\n".join([str(obj.get("page_content", "")) for obj in data])
    text = text.replace("\r", "")
    # Regex: capture question number, question block (non-greedy), and the numeric answer in parentheses
    pat = re.compile(r"(?:^|\n)\s*(\d{1,3})\.\s*(.*?)(?:\n\s*Answer\s*\((\d+)\))", re.S)
    out = []
    used_ids = set()
    for i, m in enumerate(pat.finditer(text)):
        qn = m.group(1) or str(i)
        body = (m.group(2) or "").strip()
        # Truncate at a solution marker if present to keep the stem/options only
        body = body.split("\nSol.")[0].strip()
        # Normalize whitespace
        body = re.sub(r"[ \t]+", " ", body)
        ans_num = m.group(3)
        # Build a stable id even if question numbers repeat across sections
        uid = f"neet:{qn}:{i}"
        if uid in used_ids:
            continue
        used_ids.add(uid)
        # Keep the whole body as the question (includes options like (1)...(4) when present)
        out.append({
            "id": uid,
            "question": body,
            "context": "",
            "answer": f"({ans_num})"
        })
    return out

# 2) Otherwise, strictly load from ANEETA repo's Processed Data
if not rows:
    if not repo_processed.exists():
        raise SystemExit(f"ANEETA processed data folder not found at: {repo_processed}")
    print("Using ANEETA repo processed data at:", repo_processed)

    # First, try specialized parse from solved_question_papers.json for MCQ Q/A pairs
    solved_path = repo_processed / "solved_question_papers.json"
    if solved_path.exists():
        parsed = parse_solved_mcqs(solved_path)
        if parsed:
            rows = parsed
            source_used = str(solved_path)

    # If still empty, fall back to generic extraction from other processed files (context only)
    if not rows:
        candidates = [
            "mentor_data.json",
            "processed_biology_chunks.json",
            "processed_chemistry_chunks.json",
            "processed_physics_chunks.json",
        ]
        def load_json(path: Path):
            try:
                data = json.loads(path.read_text(encoding='utf-8'))
                if isinstance(data, dict):
                    # flatten dict-of-lists if needed
                    flat = []
                    for v in data.values():
                        if isinstance(v, list):
                            flat.extend(v)
                    return flat or [data]
                return data if isinstance(data, list) else [data]
            except Exception:
                return []
        def get_ci(d: dict, keys: list[str]):
            # case-insensitive get
            lk = {k.lower(): k for k in d.keys()}
            for k in keys:
                if k.lower() in lk:
                    return d.get(lk[k.lower()])
            return None
        extracted = []
        used_file = None
        for name in candidates:
            f = repo_processed / name
            if not f.exists():
                continue
            data = load_json(f)
            for i, rec in enumerate(data):
                if not isinstance(rec, dict):
                    continue
                # These processed files generally have 'page_content' text chunks
                ctx = get_ci(rec, ["page_content","content","text","chunk"]) or ""
                if isinstance(ctx, (dict, list)):
                    ctx = json.dumps(ctx, ensure_ascii=False)
                if isinstance(ctx, str) and len(ctx.strip()) >= 20:
                    # Use the chunk as context and prompt the agent with a generic instruction
                    q_text = ctx.strip().split("\n")[0]
                    extracted.append({
                        "id": f"{name}:{i}",
                        "question": q_text[:500],
                        "context": ctx[:2000],
                        "answer": ""
                    })
            if extracted:
                rows = extracted
                used_file = f
                break
        source_used = str(used_file) if used_file else source_used

# Build DataFrame
import pandas as pd

df = pd.DataFrame(rows)
if df.empty:
    raise SystemExit("Empty dataset after parsing ANEETA data.")
print("Loaded rows:", len(df), "from:", source_used)

# Normalize columns and order
df = df.rename(columns={"sample_id":"id"})
for col in ["id","question","context","answer"]:
    if col not in df.columns:
        df[col] = None
# Cap dataset size for quick iteration
max_n = int(os.getenv("ANEETA_MAX_RECORDS", "250"))
df = df[["id","question","context","answer"]].head(max_n)
df.head(3)
```

## Baseline ANEETA Doubt Solver (non‑DSPy)
The following cells call ANEETA’s mcq_question_solver_agent through the LangGraph workflow to provide a baseline Doubt Solver run without DSPy.

```python
# Configure ANEETA environment and import ACTUAL MCQ solver agent
import os, sys, re

# Force baseline model for NEET evaluation (override any environment defaults)
os.environ["LLM_MODEL"] = "llama3.1:8b"
os.environ["CREATIVE_LLM_MODEL"] = "llama3.1:8b"
# Embeddings are required to open the persisted Chroma vector stores; use the model that matches the DBs
os.environ.setdefault("EMBEDDING_MODEL", "nomic-embed-text")
# Use the ANEETA repo's persisted Chroma DBs
os.environ.setdefault("VECTORDB_BASE_PATH", str((aneeta_dir / "src" / "aneeta" / "vectordb").resolve()))

print("LLM_MODEL:", os.environ["LLM_MODEL"]) 
print("CREATIVE_LLM_MODEL:", os.environ["CREATIVE_LLM_MODEL"]) 
print("EMBEDDING_MODEL:", os.environ["EMBEDDING_MODEL"])
print("ANEETA VECTORDB_BASE_PATH:", os.environ["VECTORDB_BASE_PATH"])

# Add ANEETA src to path for imports
sys.path.insert(0, str((aneeta_dir / "src").resolve()))
print("PYTHONPATH +:", (aneeta_dir/"src").resolve())

# Import ANEETA's actual MCQ solver agent and dependencies
from aneeta.state.models import State
from aneeta.nodes.agents import mcq_question_solver_agent as aneeta_mcq_agent
from aneeta.core.resources import llm  # Get ANEETA's configured LLM
from langchain_core.messages import HumanMessage  # Required for State messages

print("✓ Imported ANEETA's actual mcq_question_solver_agent from nodes/agents.py")

# Define a simple whitespace normalizer to clean outputs
def normalize_whitespace(text: str) -> str:
    t = text.replace("\r", "")
    # Collapse 3+ newlines to 2
    t = re.sub(r"\n{3,}", "\n\n", t)
    # Collapse runs of spaces
    t = re.sub(r" {2,}", " ", t)
    return t.strip()

# Wrapper function to adapt ANEETA's State-based agent to simple question/answer interface
def mcq_question_solver_agent(question: str, model_name: str = None, language: str = "English"):
    """
    Wrapper around ANEETA's actual mcq_question_solver_agent.
    Converts simple question string to State object and streams back response.
    
    Args:
        question: The NEET MCQ question to solve
        model_name: Optional model name to use (defaults to LLM_MODEL env var)
        language: Language for explanation (default: English)
    
    Returns:
        dict with 'text' key containing the full answer
    """
    try:
        # If model_name specified, temporarily update environment and reinitialize LLM
        original_model = os.environ.get("LLM_MODEL")
        if model_name:
            os.environ["LLM_MODEL"] = model_name
            # Force reload of LLM resource with new model
            from aneeta.core import resources
            from aneeta.llm.llm import get_llm
            resources.llm = get_llm()
        
        # Build State object with proper HumanMessage (required by ANEETA's agent)
        state = State(
            messages=[HumanMessage(content=question)],
            user_explanation_language=language,
            agent_routing="mcq_question_solver",
            teacher_vectordb_routing="biology",  # default, not used by MCQ solver
            response_stream=None
        )
        
        # Call ANEETA's actual agent
        result_state = aneeta_mcq_agent(state)
        
        # Restore original model
        if model_name:
            os.environ["LLM_MODEL"] = original_model
            # Restore original LLM
            resources.llm = get_llm()
        
        # Collect streamed response
        response_chunks = []
        for chunk in result_state.get("response_stream", []):
            response_chunks.append(str(chunk))
        
        full_response = "".join(response_chunks)
        return {"text": normalize_whitespace(full_response)}
        
    except Exception as e:
        import traceback
        return {"text": f"[MCQ solver error: {e}]\n{traceback.format_exc()[:300]}"}

# Smoke test
test_q = "Which organelle is the powerhouse of the cell? Options: (A) Ribosome (B) Mitochondria (C) Golgi (D) Lysosome"
print("\n" + "="*60)
print("Smoke Test: ANEETA's Actual MCQ Question Solver Agent")
print("="*60)
print(f"Question: {test_q}")
print("\nAnswer:")
result = mcq_question_solver_agent(test_q)
print(result.get("text", ""))
print("="*60)
```

## MCQ Solver Agent Evaluation

This section evaluates ANEETA's **MCQ Question Solver Agent** on NEET exam questions. The agent is a specialist in biology, chemistry, and physics, designed to:

1. **Answer MCQ questions** with clear explanations
2. **Provide step-by-step reasoning** to help students understand solutions
3. **Support multilingual explanations** (English + user language)

### Agent Architecture
The MCQ solver agent uses:
- **System Prompt**: Specialized instructions for NEET MCQ solving
- **Streaming Output**: Real-time response generation
- **LangChain Integration**: Prompt templates + ChatOllama + Output parsers

### Evaluation Metrics
Each model is evaluated on **3 metrics**:
1. **Fact-Check Score (0-10)**: Correctness against ground truth answers
2. **Quality Score (1-10)**: GPT-4o-mini judge assessing clarity, reasoning, and approach
3. **Latency (ms)**: Response time for generating answers

### Models Tested
- `llama3.1:8b` - Baseline model (tools-capable)
- `gemma2:9b` - Higher quality candidate
- `mistral-nemo:12b` - Advanced candidate (heavier)


```python
# Comprehensive Evaluation: 3 Models on 3 NEET MCQs using MCQ Solver Agent
import os, numpy as np, pandas as pd, time, requests, json, re
from pathlib import Path
try:
    from IPython.display import display, Markdown
except Exception:
    display = lambda x: print(x)  # fallback
    Markdown = lambda x: x

# Quick reachability check to avoid confusing errors
def _ollama_ok(url: str) -> bool:
    try:
        r = requests.get(url, timeout=5)
        return r.status_code in (200, 404)
    except Exception:
        return False

if not _ollama_ok(OLLAMA_URL):
    print("Ollama not reachable at", OLLAMA_URL)
    print("Start the service in a PowerShell:")
    print("ollama serve    # then rerun this cell")
else:
    # Helper: get available Ollama tags to filter the test set
    available = set()
    try:
        r = requests.get(f"{OLLAMA_URL}/api/tags", timeout=10)
        if r.ok:
            for m in r.json().get('models', []):
                if 'name' in m:
                    available.add(m['name'])
    except Exception:
        pass

    # Shortlist for one-click comparisons
    model_shortlist = ["llama3.1:8b", "gemma2:9b", "mistral-nemo:12b"]
    models_to_test = [m for m in model_shortlist if (not available) or (m in available)]
    missing = [m for m in model_shortlist if m not in models_to_test]
    if missing:
        print("Missing models not found in Ollama; pull as needed:")
        for m in missing:
            print(f" - ollama pull {m}")
    if not models_to_test:
        models_to_test = ["llama3.1:8b"]  # safe default

    # Load real questions from ANEETA solved question papers
    print("\n" + "="*70)
    print("Loading NEET MCQs from solved_question_papers.json...")
    print("="*70)
    solved_papers_path = aneeta_dir / "Processed Data" / "solved_question_papers.json"
    mcq_examples = []
    
    if solved_papers_path.exists():
        try:
            data = json.loads(solved_papers_path.read_text(encoding='utf-8'))
            # Build a simple question -> answer mapping
            text = "\n\n".join([str(obj.get("page_content", "")) for obj in data])
            # Extract questions and their answers
            pattern = re.compile(r"(\d{1,3})\.\s*(.*?)(?:\n\s*Answer\s*\((\d+)\))", re.S)
            
            # Collect all Q&A pairs
            all_questions = []
            for match in pattern.finditer(text):
                q_num = match.group(1)
                q_body = match.group(2).strip().split("\nSol.")[0].strip()
                answer_num = match.group(3)
                
                # Normalize whitespace
                q_body = re.sub(r'\s+', ' ', q_body).strip()
                
                # Determine subject from question content (simple heuristic)
                subject = "General"
                if any(word in q_body.lower() for word in ['force', 'velocity', 'acceleration', 'mass', 'energy', 'field', 'charge']):
                    subject = "Physics"
                elif any(word in q_body.lower() for word in ['reaction', 'mol', 'compound', 'element', 'bond', 'solution']):
                    subject = "Chemistry"
                elif any(word in q_body.lower() for word in ['cell', 'dna', 'gene', 'organism', 'enzyme', 'protein']):
                    subject = "Biology"
                
                all_questions.append({
                    "question_number": q_num,
                    "subject": subject,
                    "question": q_body,
                    "correct_answer": f"({answer_num})"
                })
            
            # Select 3 diverse questions (try to get one from each subject)
            physics_q = [q for q in all_questions if q['subject'] == 'Physics']
            chemistry_q = [q for q in all_questions if q['subject'] == 'Chemistry']
            biology_q = [q for q in all_questions if q['subject'] == 'Biology']
            
            if physics_q:
                mcq_examples.append(physics_q[0])
            if chemistry_q:
                mcq_examples.append(chemistry_q[0])
            if biology_q:
                mcq_examples.append(biology_q[0])
            
            # If we don't have 3 yet, fill with any questions
            while len(mcq_examples) < 3 and len(all_questions) > len(mcq_examples):
                for q in all_questions:
                    if q not in mcq_examples:
                        mcq_examples.append(q)
                        break
            
            print(f"✓ Loaded {len(mcq_examples)} questions from solved papers\n")
            for i, q in enumerate(mcq_examples, 1):
                print(f"Q{i} ({q['subject']}): {q['question'][:80]}...")
                print(f"    Correct Answer: {q['correct_answer']}\n")
                
        except Exception as e:
            print(f"⚠ Could not load questions: {e}")

    # METRIC 1: Fact-Check Function (Correctness against ground truth)
    def fact_check_answer(model_answer: str, correct_answer: str) -> int:
        """
        Compares model answer against ground truth from solved_question_papers.json.
        Returns score 0-10 based on answer correctness.
        """
        # Extract the option from model answer (look for (A), (B), (C), (D), (1), (2), (3), (4))
        model_option = re.search(r"\(([A-D1-4])\)", model_answer)
        model_option = model_option.group(1) if model_option else None
        
        correct_option = re.search(r"\(([A-D1-4])\)", correct_answer)
        correct_option = correct_option.group(1) if correct_option else None
        
        if model_option and correct_option:
            return 10 if model_option == correct_option else 0
        else:
            # Fuzzy match: check if correct answer appears in response
            return 7 if correct_answer.lower() in model_answer.lower() else 2

    # METRIC 2: LLM Judge Function (Quality assessment via GPT-4o-mini)
    def judge_answer_quality(question: str, answer: str) -> dict:
        """
        Uses GPT-4o-mini to judge answer quality on clarity, reasoning steps, and explanation quality.
        Returns score 1-10 and reasoning.
        """
        try:
            # Check if OpenAI API key is set
            openai_key = os.getenv("OPENAI_API_KEY")
            if not openai_key or openai_key == "your-api-key-here":
                return {"score": 5, "reasoning": "OpenAI API key not configured - set in cell 13"}
            
            from langchain_openai import ChatOpenAI
            judge_llm = ChatOpenAI(model="gpt-4o-mini", api_key=openai_key, temperature=0)
            
            judge_prompt = f"""You are evaluating a student's answer to a NEET exam question for an educational AI system. Rate the answer quality on a scale of 1-10 based on THREE criteria with specific weights.

===================================================================================
CRITERIA 1: CLARITY OF EXPLANATION (Weight: 30%)
===================================================================================
Definition: 
How easy is it for a NEET student to understand the explanation? Evaluates whether the language is accessible, technical terms are explained, the flow is logical, and the presentation helps learning.

GOOD Example (Score 8-10):
"The correct answer is (B) Mitochondria. Mitochondria are called the 'powerhouse of the cell' because they produce ATP (adenosine triphosphate), which is the main energy currency used by cells. Think of ATP like a rechargeable battery that powers all cellular activities. The mitochondria have a special folded inner membrane called cristae, which increases the surface area for energy production through a process called cellular respiration."

Why it's good: Uses analogies ("rechargeable battery"), explains technical terms (ATP, cristae), flows logically from answer → function → mechanism, accessible language for students.

BAD Example (Score 1-3):
"Answer is B. Mitochondria does oxidative phosphorylation via ETC complexes I-IV and ATP synthase utilizing proton-motive force across inner membrane."

Why it's bad: No explanation of terms, assumes advanced knowledge, jargon-heavy, no context for why this matters, not helpful for learning.

===================================================================================
CRITERIA 2: LOGICAL REASONING & STEP-BY-STEP THINKING (Weight: 40%)
===================================================================================
Definition:
Does the answer break down the problem into clear, sequential steps? Can a student follow the logical flow from problem to solution? Are intermediate reasoning steps shown, not just the final answer?

GOOD Example (Score 8-10):
"Let's solve this step by step:
Step 1: Identify what we're looking for - an enzyme that joins DNA fragments
Step 2: Recall that Okazaki fragments are short DNA pieces on the lagging strand
Step 3: Consider each option:
   - DNA polymerase: Synthesizes DNA but doesn't join fragments
   - DNA ligase: Creates phosphodiester bonds between fragments ✓
   - Helicase: Unwinds DNA
   - Primase: Makes RNA primers
Step 4: DNA ligase is the only enzyme that seals gaps between DNA segments
Therefore, the answer is (B) DNA ligase."

Why it's good: Clear numbered steps, shows reasoning process, eliminates wrong options with explanations, builds to conclusion logically.

BAD Example (Score 1-3):
"The answer is DNA ligase because it joins Okazaki fragments. So (B) is correct."

Why it's bad: No breakdown of reasoning, jumps to answer, doesn't explain why other options are wrong, no intermediate thinking shown.

===================================================================================
CRITERIA 3: CORRECTNESS OF APPROACH (Weight: 30%)
===================================================================================
Definition:
Even if the final answer is wrong, is the methodology sound? Are the right scientific concepts, formulas, or biological processes being applied? Is the approach appropriate for this type of NEET question?

GOOD Example (Score 8-10):
"To find the wavelength, we use the de Broglie equation: λ = h/p, where h is Planck's constant and p is momentum.
Given: velocity v = 2×10⁶ m/s, mass of electron m = 9.1×10⁻³¹ kg
First, calculate momentum: p = mv = (9.1×10⁻³¹)(2×10⁶) = 1.82×10⁻²⁴ kg⋅m/s
Then: λ = (6.63×10⁻³⁴)/(1.82×10⁻²⁴) = 3.64×10⁻¹⁰ m
Even if there's a calculation error, this approach using de Broglie's equation is correct for particle wavelength problems."

Why it's good: Correct formula for the problem type, proper units, systematic calculation, appropriate physics concepts applied.

BAD Example (Score 1-3):
"We need wavelength, so let's use E = mc². The energy is (9.1×10⁻³¹)(3×10⁸)² = 8.19×10⁻¹⁴ J. Then wavelength = E/h = 1.2×10¹⁹ m."

Why it's bad: Wrong formula for the problem (E=mc² is for mass-energy equivalence, not de Broglie wavelength), incorrect relationship between energy and wavelength, fundamentally flawed approach.

===================================================================================

Now evaluate this answer:

Question: {question}

Student's Answer:
{answer}

===================================================================================
Provide your evaluation in this exact format:
Score: [number 1-10]
Reasoning: [3-4 sentences explaining how the answer performs on each of the three criteria: clarity (30%), logical reasoning (40%), and correctness of approach (30%). Be specific about strengths and weaknesses for each criterion.]

Your evaluation:"""
            
            response = judge_llm.invoke(judge_prompt)
            response_text = response.content if hasattr(response, 'content') else str(response)
            
            # Extract score
            score_match = re.search(r"Score:\s*(\d+)", response_text)
            score = int(score_match.group(1)) if score_match else 5
            score = max(1, min(10, score))  # Clamp to 1-10
            
            # Extract reasoning
            reasoning_match = re.search(r"Reasoning:\s*(.+)", response_text, re.S)
            reasoning = reasoning_match.group(1).strip()[:400] if reasoning_match else "No reasoning provided"
            
            return {"score": score, "reasoning": reasoning}
        except Exception as e:
            return {"score": 5, "reasoning": f"Judge error: {str(e)[:100]}"}

    # Model Evaluation Function
    def evaluate_model_on_mcqs(model_name: str):
        """
        Evaluate a single model on all 3 MCQ questions using the MCQ solver agent.
        Returns comprehensive results including all 3 metrics.
        """
        results = []
        latencies = []
        
        for idx, mcq in enumerate(mcq_examples, 1):
            print(f"\n{'='*70}")
            print(f"  QUESTION {idx}/{len(mcq_examples)}: {mcq['subject']}")
            print(f"{'='*70}")
            print(f"Q: {mcq['question'][:200]}...")
            print(f"Correct Answer: {mcq['correct_answer']}")
            print(f"\nModel: {model_name}")
            print("-"*70)
            
            # METRIC 3: Measure latency
            t0 = time.time()
            out = mcq_question_solver_agent(
                question=mcq["question"],
                model_name=model_name,
                language="English"
            )
            dt = (time.time()-t0)*1000
            latencies.append(dt)
            
            answer_text = out.get("text","")
            
            # Show full model output
            print("MODEL OUTPUT:")
            print(answer_text if answer_text else "[no response]")
            print("-"*70)
            
            # METRIC 1: Fact-check score (compare against ground truth)
            fact_score = fact_check_answer(answer_text, mcq["correct_answer"])
            
            # METRIC 2: Quality score (GPT-4o-mini judge)
            if answer_text and len(answer_text) > 10:
                quality_result = judge_answer_quality(mcq["question"], answer_text)
                quality_score = quality_result["score"]
                quality_reasoning = quality_result["reasoning"]
            else:
                quality_score = 1
                quality_reasoning = "No meaningful answer provided"
            
            # Print all 3 metrics
            print(f"METRICS:")
            print(f"  ✓ METRIC 1 - Fact-Check Score: {fact_score}/10 (Correctness vs Ground Truth)")
            print(f"  ✓ METRIC 2 - Quality Score: {quality_score}/10 (Clarity, Reasoning, Approach)")
            print(f"  ✓ METRIC 3 - Latency: {dt:.0f}ms (Response Time)")
            print(f"\nQuality Judge Reasoning:")
            print(f"  {quality_reasoning}")
            
            results.append({
                "subject": mcq["subject"],
                "question": mcq["question"][:100],
                "answer": answer_text[:500] if answer_text else "[no response]",
                "latency_ms": round(dt, 1),
                "fact_check_score": fact_score,
                "quality_score": quality_score,
                "quality_reasoning": quality_reasoning
            })
        
        return {
            "model": model_name,
            "avg_latency_ms": round(np.mean(latencies), 1),
            "avg_fact_score": round(np.mean([r["fact_check_score"] for r in results]), 1),
            "avg_quality_score": round(np.mean([r["quality_score"] for r in results]), 1),
            "results": results
        }

    # Main Evaluation Loop
    all_results = []
    print("\n" + "="*70)
    print(" COMPREHENSIVE EVALUATION")
    print("="*70)
    print(f"Agent: MCQ Question Solver Agent (ANEETA)")
    print(f"Models: {', '.join(models_to_test)}")
    print(f"Questions: {len(mcq_examples)} NEET MCQs from solved papers")
    print(f"\nEvaluation Metrics:")
    print(f"  1) Fact-Check Score (0-10): Correctness against ground truth")
    print(f"  2) Quality Score (1-10): GPT-4o-mini judge (clarity + reasoning + approach)")
    print(f"  3) Latency (ms): Response time")
    
    for m in models_to_test:
        print(f"\n{'#'*70}")
        print(f"# TESTING MODEL: {m}")
        print(f"{'#'*70}")
        eval_result = evaluate_model_on_mcqs(m)
        all_results.append(eval_result)
    
    # Build detailed DataFrame
    rows = []
    for result in all_results:
        for i, mcq_result in enumerate(result["results"], 1):
            rows.append({
                "Model": result["model"],
                "Question": f"Q{i} ({mcq_result['subject']})",
                "Answer_Preview": mcq_result["answer"][:150] + "...",
                "Latency_ms": mcq_result["latency_ms"],
                "Fact_Score": mcq_result["fact_check_score"],
                "Quality_Score": mcq_result["quality_score"],
            })
    
    df_detailed = pd.DataFrame(rows)
    
    # Summary DataFrame
    summary_rows = []
    for result in all_results:
        summary_rows.append({
            "Model": result["model"],
            "Avg_Latency_ms": result["avg_latency_ms"],
            "Avg_Fact_Score": result["avg_fact_score"],
            "Avg_Quality_Score": result["avg_quality_score"],
            "Overall_Score": round((result["avg_fact_score"] + result["avg_quality_score"]) / 2, 1)
        })
    
    df_summary = pd.DataFrame(summary_rows).sort_values("Overall_Score", ascending=False)
    
    # Optional MLflow logging
    try:
        import mlflow
        mlflow.set_experiment("aneeta-mcq-solver-agent-evaluation")
        for result in all_results:
            with mlflow.start_run(run_name=f"mcq_solver_{result['model']}"):
                mlflow.log_params({
                    "model": result["model"],
                    "agent": "mcq_question_solver_agent",
                    "num_questions": len(mcq_examples)
                })
                mlflow.log_metrics({
                    "avg_latency_ms": result["avg_latency_ms"],
                    "avg_fact_score": result["avg_fact_score"],
                    "avg_quality_score": result["avg_quality_score"],
                    "overall_score": round((result["avg_fact_score"] + result["avg_quality_score"]) / 2, 1)
                })
        print("\n✓ Logged to MLflow")
    except Exception:
        pass
    
    # Final Summary
    print("\n" + "="*70)
    print(" FINAL SUMMARY: Model Performance Comparison")
    print("="*70)
    display(df_summary)
    
    print("\n" + "="*70)
    print(" DETAILED RESULTS: All Questions")
    print("="*70)
    display(df_detailed)
    
    print("\n" + "="*70)
    print("Evaluation Complete!")
    print("="*70)
```
